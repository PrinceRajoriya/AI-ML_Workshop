```python
##Feature Engineering
import numpy as np
import pandas as pd

df = pd.read_csv('ecommerce_raw.csv')

# 1. Create Gross_Amount feature using NumPy multiplication
df["Gross_Amount"] = np.multiply(df["Product_Price"], df["Quantity"])

# 2. Calculate the total gross sales generated by all orders using NumPy sum
total_gross = np.sum(df["Gross_Amount"])

print(f"Total Gross Amount: ${total_gross:,.2f}")

```

```python
# 1. Calculate Discount_Amount using NumPy
df["Discount_Amount"] = np.multiply(df["Gross_Amount"], df["Discount"])

# 2. Calculate Final_Amount using NumPy subtraction
df["Final_Amount"] = np.subtract(df["Gross_Amount"], df["Discount_Amount"])

# 3. Find the product category with the highest total Final_Amount using Pandas
category_totals = df.groupby("Product_Category")["Final_Amount"].sum()
highest_category = category_totals.idxmax()
highest_value = category_totals.max()

print(f"Product Category with highest: {highest_category}")
print(f"Total Revenue from this category: ${highest_value:,.2f}")
```

```python
##EDA Exploratory data analysis
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Show simple descriptive statistics
print(df["Age"].describe())

# 2. Plot the histogram
plt.figure(figsize=(6, 4))
sns.histplot(df["Age"], kde=True, color="skyblue", bins=20)
plt.title("Age Distribution of Customers")
plt.xlabel("Age")
plt.ylabel("Number of Customers")
plt.show()
```

```python
# 1. Count total orders per category
order_counts = df["Product_Category"].value_counts()
print(order_counts)

# 2. Plot the bar chart
plt.figure(figsize=(7, 4))
order_counts.plot(kind="bar", color="salmon")
plt.title("Number of Orders by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Number of Orders")
plt.xticks(rotation=45)
plt.show()

# Find the winner
print(f"Category with the highest order volume: {order_counts.idxmax()}")
```
