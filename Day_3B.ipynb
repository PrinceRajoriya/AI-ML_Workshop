```python
#Data preprocessing
import pandas as pd
import numpy as np

df = pd.read_csv('ecommerce_raw.csv')

print(df.isnull().sum())
```

```python

df["Signup_Date"] = pd.to_datetime(df["Signup_Date"])
df["Order_Date"] = pd.to_datetime(df["Order_Date"])

# 2. Check the data type to verify it worked
print(df[["Signup_Date", "Order_Date"]].dtypes)
```

```python

# 1. Pull out the month and day names
df["Month_Name"] = df["Order_Date"].dt.month_name()
df["Day_Name"] = df["Order_Date"].dt.day_name()

# 2. Find which month and day appear the most
busiest_month = df["Month_Name"].value_counts().idxmax()
busiest_day = df["Day_Name"].value_counts().idxmax()

print("Busiest Month:", busiest_month)
print("Busiest Day:", busiest_day)
```

```python
# 1. Find columns that hold text words
text_columns = df.select_dtypes(include=["object"]).columns
print("Text columns found:", text_columns)

# 2. Convert text columns to numbers (One-Hot Encoding)
df_encoded = pd.get_dummies(df, columns=text_columns)

# 3. See how many columns we started with vs. how many we have now
print("Original Columns:", df.shape[1])
print("New Columns after encoding:", df_encoded.shape[1])
```

```python
# 1. Pick a column with big numbers (example: 'Income')
# 2. Scale it so values match a standard scale (Z-score scaling)
import pandas as pd
import numpy as np

price_mean = np.mean(df_encoded["Product_Price"])
price_std = np.std(df_encoded["Product_Price"])

df_encoded["Income_Scaled"] = (df_encoded["Product_Price"] - price_mean) / price_std

# 3. Compare values before and after
print("Before Scaling (Example values):\n", df_encoded["Product_Price"].head())
print("\nAfter Scaling (Centered around 0):\n", df_encoded["Income_Scaled"].head())
```
