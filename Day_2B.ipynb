```python
import pandas as pd
import numpy as np

```

```python
df = pd.read_csv("ecommerce_sales.csv")
print(df)
```

```python
#convert this into numpy array
quantity_array = df['Quantity'].values   #you use also use[to_numpy() = values]
#print this numpy array
print(quantity_array)

```

```python
mean_Quantity = df['Quantity'].mean()
max_Quantity = df['Quantity'].max()
min_Quantity = df['Quantity'].min()

# Print 
print(f"Mean Quantity: {mean_Quantity}")
print(f"Maximum Quantity: {max_Quantity}")
print(f"Minimum Quantity: {min_Quantity}")
```

```python
mean_price = df['Price'].mean()
std_price = df['Price'].std()

print(f"Mean Price: {mean_price}")
print(f"Standard Deviation of Price: {std_price}")

#print(df['price'].describe()) quick check for mean,std, min,max,25%,50%,75% and count
```

```python
revenue = df['Quantity'] * df['Price']
print(revenue)
print(f"Revenue: {revenue.sum()}")
```

```python
top3 = revenue.nlargest(3)
print(top3)
```
